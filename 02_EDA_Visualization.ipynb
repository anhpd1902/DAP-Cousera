# SpaceX - Exploratory data analysis

Show launches and landing outcomes by year, site and payload mass.

**Scope:** Independent reproduction from a reconstructed public 90-record sample; not a record of the learner's original IBM lab execution.


```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("spacex_sample_90_reconstructed.csv")
display(df.head())
```

```python
fig,ax=plt.subplots(figsize=(9,4))
(df.groupby("Year")["Class"].mean()*100).plot(ax=ax,marker="o")
ax.set(xlabel="Launch year",ylabel="Landing-success share (%)",title="Historical label share by year")
plt.tight_layout(); plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8,4))
(df.groupby("LaunchSite")["Class"].mean()*100).plot.bar(ax=ax)
ax.set(ylabel="Success share (%)",title="Outcome by launch site")
plt.xticks(rotation=15); plt.tight_layout(); plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8,4))
for label,sub in df.groupby("Class"):
    ax.scatter(sub.FlightNumber,sub.PayloadMass_kg,label=f"Class {label}",alpha=.75)
ax.set(xlabel="Flight Number",ylabel="Payload mass (kg)",title="Flight history and payload mass")
ax.legend(); plt.show()
```
