# SpaceX - SQL exploratory analysis

Use SQLite in memory to reproduce aggregate queries over the selected-column dataset.

**Scope:** Independent reproduction from a reconstructed public 90-record sample; not a record of the learner's original IBM lab execution.


```python
import pandas as pd
import sqlite3
df=pd.read_csv("spacex_sample_90_reconstructed.csv")
conn=sqlite3.connect(":memory:")
df.to_sql("SPACEX",conn,index=False,if_exists="replace")
```

```python
display(pd.read_sql_query("""SELECT LaunchSite,COUNT(*) AS Launches,SUM(Class) AS SuccessfulLandings,
ROUND(AVG(Class)*100,1) AS SuccessPct FROM SPACEX GROUP BY LaunchSite ORDER BY Launches DESC""",conn))
```

```python
display(pd.read_sql_query("""SELECT Year,COUNT(*) AS Launches,SUM(Class) AS SuccessfulLandings,
ROUND(AVG(Class)*100,1) AS SuccessPct FROM SPACEX GROUP BY Year ORDER BY Year""",conn))
```

```python
display(pd.read_sql_query("""SELECT Orbit,COUNT(*) AS Launches,
ROUND(AVG(PayloadMass_kg),1) AS MeanPayloadMassKg FROM SPACEX GROUP BY Orbit ORDER BY Launches DESC""",conn))
```
